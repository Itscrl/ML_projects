# Bike-sharing linear regression

An undergraduate regression exercise using the daily UCI Bike Sharing data. It predicts `cnt` from the `season` index with an 80/20 random split and `random_state=1`.

This is a simple teaching baseline: season is categorical, and treating its category index as a continuous feature is restrictive. A random split is not a future-date forecasting evaluation. See `data/README.md` for data attribution and setup.


```python
# Import the libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.metrics import mean_squared_error
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
```

```python
from pathlib import Path

repo_root = next(
    (p for p in (Path.cwd(), *Path.cwd().parents)
     if (p / "notebooks/bike_sharing_linear_regression.ipynb").is_file()),
    None,
)
if repo_root is None:
    raise FileNotFoundError("Open this notebook from the repository or notebooks directory.")
data_path = repo_root / "data/raw/bike-sharing/day.csv"
if not data_path.is_file():
    raise FileNotFoundError("Run python scripts/download_bike_sharing.py from the repository root.")

dataset = pd.read_csv(data_path)
X = dataset["season"].to_numpy()
y = dataset["cnt"].to_numpy()
print(f"Daily records: {len(dataset)}")
```

```python
# split data into training set and test set

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = .20, random_state = 1)

print(X_train.shape)
print(y_train.shape)
```

```python
# Fitting simple linear regression to training set 
model = LinearRegression()
model.fit(X_train.reshape(-1, 1), y_train)
```

```python
# Predicting the test set results
y_pred = model.predict(X_test.reshape(-1, 1))
```

```python
# Visualizing the training set
plt.scatter(X_train, y_train, color = 'red')
plt.plot(X_train, model.predict(X_train.reshape(-1, 1)), color = 'blue')
plt.title('Linear Regression Train Set')
plt.xlabel('Season index (dataset coding)')
plt.ylabel('Daily rentals')
plt.show()
```

```python
# Visualizing the test set 
plt.scatter(X_test, y_test, color = 'red')
mse = mean_squared_error(y_test, y_pred)
baseline_pred = np.full(y_test.shape, y_train.mean(), dtype=float)
baseline_mse = mean_squared_error(y_test, baseline_pred)
print(f'Test MSE: {mse:.2f}')
print(f'Training-mean baseline MSE: {baseline_mse:.2f}')
plt.plot(X_test, y_pred, color = 'blue')
plt.title('Linear Regression Test Set')
plt.xlabel('Season index (dataset coding)')
plt.ylabel('Daily rentals')
plt.show()
```
